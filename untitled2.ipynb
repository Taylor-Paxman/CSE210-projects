```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from matplotlib.ticker import MultipleLocator
from sklearn.model_selection import train_test_split
import xgboost as xgb
from sklearn.model_selection import GroupShuffleSplit, GroupKFold, RandomizedSearchCV
from sklearn.metrics import mean_squared_error, r2_score
from scipy.stats import randint, uniform, loguniform
from sklearn.inspection import permutation_importance
```

```python
housing_data = pd.read_csv('https://raw.githubusercontent.com/Taylor-Paxman/CSE450/refs/heads/main/housing_market.csv')
```

```python
print("Number of rows:", housing_data.shape[0])
```

```python
sns.histplot(data=housing_data, x="ListPrice", bins=50)

plt.title("Distribution of House Prices")
plt.xlabel("House Price ($)")
plt.ylabel("Number of Houses")

ax = plt.gca()

ax.xaxis.set_major_locator(MultipleLocator(1_000_000))
ax.set_xticklabels([f"${x/1_000_000:.0f}M" for x in ax.get_xticks()])
plt.tight_layout()

plt.show()
```

```python
clean_date = housing_data.copy()
clean_date["ListingDate"] = clean_date["ListingDate"].astype(str).str[:8]
```

```python
clean_date = clean_date.drop(columns=["ListingDate"])
```

```python
y = clean_date["ListPrice"]
x = clean_date.drop(columns=["ListPrice", "City", "ListingID", "Neighborhood","PropertyType","Condition"])
```

```python
x_train, x_test, y_train, y_test = train_test_split(x, y, test_size=0.2, random_state=42)
```

```python
model = xgb.XGBRegressor()
```

```python
model.fit(x_train, y_train)
```

```python
predictions = model.predict(x_test)
```

```python
rmse = np.sqrt(mean_squared_error(y_test, predictions))
```

```python
r2 = r2_score(y_test, predictions)
```

```python
print("Initial Model RMSE:", rmse)
print("Initial Model R²:", r2)
```

```python
param_grid = {
    "n_estimators": randint(100, 501),
    "max_depth": randint(3, 9),
    "learning_rate": loguniform(0.01, 0.2),
    "subsample": uniform(0.7, 0.3),
    "colsample_bytree": uniform(0.7, 0.3)
}
```

```python
xgb_model = xgb.XGBRegressor(
    random_state=42,
    objective="reg:squarederror"
)
```

```python
random_search = RandomizedSearchCV(
    estimator=xgb_model,
    param_distributions=param_grid,
    n_iter=20,
    scoring="neg_root_mean_squared_error",
    cv=3,
    random_state=42,
    n_jobs=-1,
    verbose=1
)
```

```python
random_search.fit(x_train, y_train)
```

```python
print("Best Parameters:")
print(random_search.best_params_)
```

```python
print("Best Cross-Validation RMSE:", -random_search.best_score_)
```

```python
tuned_model = random_search.best_estimator_
```

```python
tuned_predictions = tuned_model.predict(x_test)
```

```python

tuned_rmse = np.sqrt(mean_squared_error(y_test, tuned_predictions))
```

```python
tuned_rmse = np.sqrt(mean_squared_error(y_test, tuned_predictions))
```

```python
tuned_r2 = r2_score(y_test, tuned_predictions)
```

```python
print("Tuned Model RMSE:", tuned_rmse)
print("Tuned Model R²:", tuned_r2)
```

```python
comparison = pd.DataFrame({
    "Model": ["Original Model", "Tuned Model"],
    "RMSE": [rmse, tuned_rmse],
    "R²": [r2, tuned_r2]
})

print(comparison.to_string(index=False))
```

```python
importance = pd.Series(tuned_model.feature_importances_, index=x_train.columns).sort_values()

plt.figure(figsize=(8, 6))
importance.plot.barh()
plt.title("Feature Importance (Tuned XGBoost Model)")
plt.xlabel("Importance")
plt.tight_layout()
plt.show()

print(importance.sort_values(ascending=False).round(3))
```

```python
perm = permutation_importance(tuned_model, x_test, y_test, scoring="r2", n_repeats=5, random_state=42)
perm_imp = pd.Series(perm.importances_mean, index=x_test.columns).sort_values()

plt.figure(figsize=(8, 6))
perm_imp.plot.barh(color="darkorange")
plt.title("Permutation Importance (drop in R² when feature is shuffled)")
plt.xlabel("Decrease in R²")
plt.tight_layout()
plt.show()

print("\nPermutation importance:")
print(perm_imp.sort_values(ascending=False).round(3))
```
